# Student Performance Analysis

In this project I analyse the exam scores of 200+ students using Pandas, NumPy and Matplotlib.

**What I did:**
- Cleaned the data using Pandas
- Found mean, median and standard deviation using NumPy
- Made graphs using Matplotlib

## Step 1: Import the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Step 2: Load the data

The marks of 200 students are in a csv file called `students_performance.csv`. Keep this file in the same folder as the notebook.

```python
df = pd.read_csv("students_performance.csv")

df.head()
```

## Step 3: Look at the data

```python
print(df.shape)
```

```python
df.info()
```

```python
df.describe()
```

## Step 4: Clean the data

### Check missing values

```python
df.isnull().sum()
```

### Check duplicates

```python
df.duplicated().sum()
```

There are some missing values and 2 duplicate rows. I will remove the duplicates and fill the missing values with the mean of that column.

```python
# remove duplicates
df = df.drop_duplicates()

# fill missing values with mean
df["Maths"] = df["Maths"].fillna(df["Maths"].mean())
df["Science"] = df["Science"].fillna(df["Science"].mean())
df["English"] = df["English"].fillna(df["English"].mean())
df["History"] = df["History"].fillna(df["History"].mean())
df["Computer"] = df["Computer"].fillna(df["Computer"].mean())

print(df.isnull().sum())
print("Rows now:", len(df))
```

### Add Total and Average columns

```python
subjects = ["Maths", "Science", "English", "History", "Computer"]

df["Total"] = df[subjects].sum(axis=1)
df["Average"] = df["Total"] / 5

df.head()
```

## Step 5: Statistics using NumPy

### Mean

```python
for sub in subjects:
    print(sub, "mean =", round(np.mean(df[sub]), 2))
```

### Median

```python
for sub in subjects:
    print(sub, "median =", round(np.median(df[sub]), 2))
```

### Standard deviation

```python
for sub in subjects:
    print(sub, "std =", round(np.std(df[sub]), 2))
```

### Putting everything in one table

```python
mean_list = []
median_list = []
std_list = []

for sub in subjects:
    mean_list.append(np.mean(df[sub]))
    median_list.append(np.median(df[sub]))
    std_list.append(np.std(df[sub]))

result = pd.DataFrame({
    "Subject": subjects,
    "Mean": mean_list,
    "Median": median_list,
    "Std": std_list
})

result
```

## Step 6: Graphs using Matplotlib

### Graph 1: Average marks of each subject

```python
plt.figure(figsize=(8, 5))
plt.bar(result["Subject"], result["Mean"], color="skyblue")
plt.xlabel("Subject")
plt.ylabel("Average Marks")
plt.title("Average Marks in Each Subject")
plt.show()
```

### Graph 2: Mean vs Median

```python
x = np.arange(len(subjects))

plt.figure(figsize=(8, 5))
plt.bar(x - 0.2, result["Mean"], 0.4, label="Mean")
plt.bar(x + 0.2, result["Median"], 0.4, label="Median")
plt.xticks(x, subjects)
plt.xlabel("Subject")
plt.ylabel("Marks")
plt.title("Mean vs Median")
plt.legend()
plt.show()
```

### Graph 3: Standard deviation of each subject

```python
plt.figure(figsize=(8, 5))
plt.plot(result["Subject"], result["Std"], marker="o", color="red")
plt.xlabel("Subject")
plt.ylabel("Standard Deviation")
plt.title("Standard Deviation of Each Subject")
plt.grid()
plt.show()
```

### Graph 4: Histogram of Maths marks

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Maths"], bins=10, color="green", edgecolor="black")
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Maths Marks Distribution")
plt.show()
```

### Graph 5: Boxplot of all subjects

```python
plt.figure(figsize=(8, 5))
plt.boxplot([df[s] for s in subjects])
plt.xticks([1, 2, 3, 4, 5], subjects)
plt.ylabel("Marks")
plt.title("Boxplot of All Subjects")
plt.show()
```

## Step 7: Conclusion

```python
best = result.loc[result["Mean"].idxmax(), "Subject"]
worst = result.loc[result["Mean"].idxmin(), "Subject"]
spread = result.loc[result["Std"].idxmax(), "Subject"]

print("Total students:", len(df))
print("Best subject:", best)
print("Weakest subject:", worst)
print("Subject with most difference in marks:", spread)
```

**What I learned**
- The best average marks are in the subject printed above and the weakest in the other one.
- A bigger standard deviation means the marks are more spread out.
- Pandas makes cleaning data easy, NumPy is good for calculations and Matplotlib makes the results easy to see.
